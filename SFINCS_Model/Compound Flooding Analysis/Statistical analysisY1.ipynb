## For three factors use this formulas!

$$A = \frac{1}{4n}\left[a - (1) + ab - b + ac - c + abc - bc\right]$$
$$B = \frac{1}{4n}\left[b + ab + bc + abc - (1) - a - c - ac\right]$$
$$C = \frac{1}{4n}\left[c + ac + bc + abc - (1) - a - b - ab\right]$$
$$AB = \frac{1}{4n}\left[ab - a - b + (1) + abc - bc - ac + c\right]$$
$$AC = \frac{1}{4n}\left[(1) - a + b - ab - c + ac - bc + abc\right]$$
$$BC = \frac{1}{4n}\left[(1) + a - b - ab - c - ac + bc + abc\right]$$
$$ABC = \frac{1}{4n}\left[abc - bc - ac + c - ab + b + a - (1)\right]$$

$$SS_A = \frac{(A * 4n)^2}{4n}$$
$$SS_B = \frac{(B * 4n)^2}{4n}$$
$$SS_C = \frac{(C * 4n)^2}{4n}$$
$$SS_{AB} = \frac{(AB * 4n)^2}{4n}$$
$$SS_{AC} = \frac{(AC * 4n)^2}{4n}$$
$$SS_{BC} = \frac{(BC * 4n)^2}{4n}$$
$$SS_{ABC} = \frac{(ABC * 4n)^2}{4n}$$

$$SS_T = SS_A + SS_B + SS_C + SS_{AB} + SS_{AC} + SS_{BC} + SS_{ABC}

```python
import pandas as pd

data = pd.read_csv('example3.csv', index_col=0, delimiter=';')
data.head()
```

```python
f = 1/8

A =  f * (data.loc['a', 'Y'] - data.loc['(-1)', 'Y'] + data.loc['ab', 'Y'] - data.loc['b', 'Y'] + data.loc['ac', 'Y'] - data.loc['c', 'Y'] + data.loc['abc', 'Y'] - data.loc['bc', 'Y'])
B =  f * (- data.loc['a', 'Y'] - data.loc['(-1)', 'Y'] + data.loc['ab', 'Y'] + data.loc['b', 'Y'] - data.loc['ac', 'Y'] - data.loc['c', 'Y'] + data.loc['abc', 'Y'] + data.loc['bc', 'Y'])
C =  f * (- data.loc['a', 'Y'] - data.loc['(-1)', 'Y'] - data.loc['ab', 'Y'] - data.loc['b', 'Y'] + data.loc['ac', 'Y'] + data.loc['c', 'Y'] + data.loc['abc', 'Y'] + data.loc['bc', 'Y'])

AB =  f * (- data.loc['a', 'Y'] - data.loc['(-1)', 'Y'] + data.loc['ab', 'Y'] - data.loc['b', 'Y'] - data.loc['ac', 'Y'] - data.loc['c', 'Y'] + data.loc['abc', 'Y'] - data.loc['bc', 'Y'])
AC =  f * (- data.loc['a', 'Y'] - data.loc['(-1)', 'Y'] - data.loc['ab', 'Y'] - data.loc['b', 'Y'] + data.loc['ac', 'Y'] - data.loc['c', 'Y'] + data.loc['abc', 'Y'] - data.loc['bc', 'Y'])
BC =  f * (- data.loc['a', 'Y'] - data.loc['(-1)', 'Y'] - data.loc['ab', 'Y'] - data.loc['b', 'Y'] - data.loc['ac', 'Y'] - data.loc['c', 'Y'] + data.loc['abc', 'Y'] + data.loc['bc', 'Y'])

ABC =  f * (- data.loc['a', 'Y'] - data.loc['(-1)', 'Y'] - data.loc['ab', 'Y'] - data.loc['b', 'Y'] - data.loc['ac', 'Y'] - data.loc['c', 'Y'] + data.loc['abc', 'Y'] - data.loc['bc', 'Y'])
```

```python
SS_A = (A * f)**2 / f
SS_B = (B * f)**2 / f
SS_C = (C * f)**2 / f
SS_AB = (AB * f)**2 / f
SS_AC = (AC * f)**2 / f
SS_BC = (BC * f)**2 / f
SS_ABC = (ABC * f)**2 / f

SS_T = SS_A + SS_B + SS_C + SS_AB + SS_AC + SS_BC + SS_ABC
```

```python
eff_A = SS_A / SS_T * 100
eff_B = SS_B / SS_T * 100
eff_C = SS_C / SS_T * 100
eff_AB = SS_AB / SS_T * 100
eff_AC = SS_AC / SS_T * 100
eff_BC = SS_BC / SS_T * 100
eff_ABC = SS_ABC / SS_T * 100

print('eff_A=', eff_A)
print('eff_B=', eff_B)
print('eff_C=', eff_C)
print('eff_AB=', eff_AB)
print('eff_AC=', eff_AC)
print('eff_BC=', eff_BC)
print('eff_ABC=', eff_ABC)
```

```python
import pandas as pd

# 1. Lees de CSV in
data = pd.read_csv("Tabel_7_9_Y1_Results.csv")
data.columns = data.columns.str.strip()

# Rond Y eventueel af op 2 decimalen
data["Y"] = data["Y"].round(2)

# 2. Zet de fysieke kolommen (+ en -) direct om naar +1 en -1
mapping = {"+": 1, "-": -1}
data["A"] = data["A (Wind)"].map(mapping)
data["B"] = data["B (Precip)"].map(mapping)
data["C"] = data["C (Surge)"].map(mapping)
data["D"] = data["D (Infilt)"].map(mapping)

# 3. Genereer alle interactiekolommen door simpel vermenigvuldigen
data["AB"] = data["A"] * data["B"]
data["AC"] = data["A"] * data["C"]
data["AD"] = data["A"] * data["D"]
data["BC"] = data["B"] * data["C"]
data["BD"] = data["B"] * data["D"]
data["CD"] = data["C"] * data["D"]
data["ABC"] = data["A"] * data["B"] * data["C"]
data["ABD"] = data["A"] * data["B"] * data["D"]
data["ACD"] = data["A"] * data["C"] * data["D"]
data["BCD"] = data["B"] * data["C"] * data["D"]
data["ABCD"] = data["A"] * data["B"] * data["C"] * data["D"]

# List van alle factoren
factors = [
    "A",
    "B",
    "C",
    "D",
    "AB",
    "AC",
    "AD",
    "BC",
    "BD",
    "CD",
    "ABC",
    "ABD",
    "ACD",
    "BCD",
    "ABCD",
]

# 4. Bereken SS en % per factor
ss_dict = {}
for factor in factors:
    contrast = (data[factor] * data["Y"]).sum()
    ss_dict[factor] = (contrast**2) / 16

ss_total = sum(ss_dict.values())

# Print het resultaat
for factor in factors:
    pct = (ss_dict[factor] / ss_total) * 100
    print(f"eff_{factor:4s} = {pct:6.2f}%")
```
