**For the Time discretization (TDIS)**

```python
# 1. Read CSV
df = pd.read_csv(r"C:\SAN_FELIPE\SAN_FELIPE\forcing\CHD&GHB\05_Corrected_total_water_level_2024_february.csv")
num_records = len(df)  # 241 hourly records (because 10 days * 24 hours/day = 240 hours)
print(num_records)


# 2. Parameters for hourly time discretization in minutes
pname = 'tdis'
time_units = 'minutes'
len_sptran = 60.0 # Length of each stress period in minutes (1 hour)
int_tran = 5      # Number of time steps per stress period (e.g., 1 step of 60 min) #more time steps within hour see .tdis

# Option for steady-state start period
include_steady_period = True 

if include_steady_period:
    nper = num_records  # 1 steady-state + 240 hourly transient periods
    perlen = np.ones(nper) * len_sptran
    nstp = np.ones(nper, dtype=int) * int_tran
    tsmult = np.ones(nper)
    
    # Settings for initial steady-state period
    perlen[0] = 1.0  # Length does not affect steady-state solution
    nstp[0] = 1
else:
    nper = num_records  # 240 transient stress periods
    
    perlen = np.ones(nper) * len_sptran
    nstp = np.ones(nper, dtype=int) * int_tran
    tsmult = np.ones(nper)

# Combine into perioddata list of tuples: (perlen, nstp, tsmult)
perioddata = list(zip(perlen, nstp, tsmult))

# Create ModflowTdis object
tdis = flopy.mf6.ModflowTdis(
    sim, 
    pname=pname, 
    time_units=time_units, 
    nper=nper, 
    perioddata=perioddata
)
print(nper)
```

**Iterative Model Solver (IMS) => I did not change anything here**

```python
#Se crea el paquete del solver IMS: Iterative Model Solver

ims6 = sim6.get_package('ims') #solucionador
pname = 'ims'
print_option = ims6.print_option.get_data()
complexity = None  #complejidad del modelo, complejo para modelos altamente no lineales, VA None por que yo defino manualmente los parametros de abajo
outer_dvclose = 0.01 #criterio de convergencia en la variable en cuestión en este caso carga hidráuica -> 1 milimetro
outer_maximum = 150 #maximo numero de iteraciones externas
under_relaxation = None
inner_maximum = 250 # maximo numero de iteraciones internas
inner_dvclose = 0.001 #se sugiere que sea un orden de magnitud menor que outer_dvclose
#rcloserecord = ims6.rcloserecord.get_data() #si no se especifica se utiliza criterio norma infinita
linear_acceleration = 'cg' #metodo del gradiente conjugado precondicionado utilizado por MF5 -> PCG
scaling_method = None #enfoque para escalar la matriz 
reordering_method = None #se mantiene el orden original de la matriz
relaxation_factor = 0.97 #un factor de 0.97 optimiza la convergencia

#Se crea el objeto ims
ims = flopy.mf6.ModflowIms(sim, pname=pname, print_option=print_option, complexity=complexity, outer_dvclose=outer_dvclose, outer_maximum=outer_maximum, 
                           under_relaxation=under_relaxation, inner_maximum=inner_maximum, inner_dvclose=inner_dvclose, 
                           linear_acceleration=linear_acceleration, scaling_method=scaling_method, reordering_method=reordering_method, 
                           relaxation_factor=relaxation_factor)

sim.register_ims_package(ims, [mf.name])
```

**Recharge package (RCH)**

```python
# include_steady_period = True  
Cr = 0.40  # Infiltration coefficient (Medina-Rosado, 2024) looks a bit high #show difference between infiltration and recharge, infiltration is lower than recharge 

df_P = pd.read_excel(r"C:\SAN_FELIPE\SAN_FELIPE\forcing\RCH\Precipitation_storm_february.xlsx")

tp_hourly_mm = df_P["tp_mm"].values
# num_records = len(tp_hourly_mm)  
print(f'length tp_hourly {len(tp_hourly_mm)}')

# nper = (num_records + 1) if include_steady_period else num_records 

r_flux_series = (Cr * (tp_hourly_mm / 1000.0)) / 60.0  # Output unit: m/min
# print(r_flux_series)
print(f'lenght r_flux {len(r_flux_series)}')

offset = 1 if include_steady_period else 0
recharge = {}

# Set initial steady-state period (Python key 0 -> MODFLOW 6 SP 1)
if include_steady_period:
    recharge[0] = 0.0

# Map all hourly flux values to their respective stress periods
for i, flux in enumerate(r_flux_series):
    sp = i + offset
    recharge[sp] = float(flux)

print(sp, flux)
print(len(recharge))

pname = "rch"
# irch = 0  # Apply recharge to top layer (Layer 0)

rch = flopy.mf6.ModflowGwfrcha(
    mf,                  
    pname=pname,
    readasarrays=True,
    save_flows=True,
    recharge=recharge,
)

recharge_keys = list(recharge.keys())

print(f'lenght recharge keys {len(recharge_keys)}')

print("Aantal perioden in RCH:", len(recharge_keys))          # Verwacht: 241
print("Lowest RCH periode index (Python):", min(recharge_keys))  # Verwacht: 0
print("Highest RCH periode index (Python):", max(recharge_keys)) # Verwacht: 240
```
